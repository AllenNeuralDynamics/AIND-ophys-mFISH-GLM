```python
data_dir = Path('/root/capsule/data/')
data_folder_names = [d.split('/')[-1] for d in glob.glob(str(data_dir) + '/*') if Path(d).is_dir()]
session_names = [d for d in data_folder_names if ('processed' not in d) and
                 ('multiplane-ophys' in d) and
                 (d!='stimuli')]
session_names
```

```python
si = 0
session_name = session_names[si]
raw_path = data_dir / session_name
processed_path = list(data_dir.glob(f'{session_name}_processed_*'))[0]
plane_folders = np.sort([str(p) for p in processed_path.glob("*") if p.is_dir() and 
                 not 'nwb' in str(p) and 
                 not 'nextflow' in str(p)])
eye_tracking_path = list(data_dir.glob(f'{session_name}_dlc-eye_*'))[0]
```
